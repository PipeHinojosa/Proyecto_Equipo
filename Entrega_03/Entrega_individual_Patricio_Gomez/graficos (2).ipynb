```python
import pandas as pd
import altair as alt

# Leer Excel
df = pd.read_excel('/content/Base de datos Patricio Gómez (Limpia).xlsx')

# Agrupar datos
df_grouped = df.groupby(['Temporada', 'Equipo'])['Valor de mercado'].sum().reset_index()

# Crear gráfico
chart = alt.Chart(df_grouped).mark_line(point=True).encode(
    x=alt.X('Temporada:O', title='Temporada'),
    y=alt.Y('Valor de mercado:Q', title='Valor de mercado total'),
    color=alt.Color('Equipo:N', title='Equipo'),
    tooltip=['Equipo', 'Temporada', 'Valor de mercado']
).properties(
    width=900,
    height=500,
    title='Evolución del valor de mercado en el fútbol chileno'
).interactive()

chart
```

```python
import pandas as pd
import altair as alt

# Leer Excel
df = pd.read_excel('/content/Base de datos Patricio Gómez (Limpia).xlsx')

# Agrupar por temporada y equipo
df_grouped = df.groupby(['Temporada', 'Equipo'])['Valor de mercado'].sum().reset_index()

# Selección interactiva
selection = alt.selection_point(fields=['Equipo'], bind='legend')

# Crear gráfico
chart = alt.Chart(df_grouped).mark_line(point=True).encode(
    x=alt.X('Temporada:O', title='Temporada'),
    y=alt.Y('Valor de mercado:Q', title='Valor de mercado total'),
    color=alt.Color('Equipo:N', title='Equipo'),
    opacity=alt.condition(selection, alt.value(1), alt.value(0.15)),
    tooltip=['Equipo', 'Temporada', 'Valor de mercado']
).add_params(
    selection
).properties(
    width=900,
    height=500,
    title='Evolución del valor de mercado en el fútbol chileno'
)

chart
```
