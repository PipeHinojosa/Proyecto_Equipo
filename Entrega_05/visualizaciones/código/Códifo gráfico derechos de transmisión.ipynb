```python
import pandas as pd
import matplotlib.pyplot as plt

# Crear datos directamente (no necesitas leer Excel)
datos = {
    "Club": [
        "Colo Colo",
        "Universidad de Chile",
        "Universidad Católica",
        "Otros clubes"
    ],

    "Dinero recibido": [
        5690237000,
        5315726000,
        4586971000,
        46300579000
    ]
}

df = pd.DataFrame(datos)

# Colores:
# Blanco = Colo Colo
# Azul = U
# Celeste = Católica
# Gris = Otros

colores = ['white', '#003DA5', '#6EC6FF', '#BDBDBD']

plt.figure(figsize=(10,10))

wedges, textos, porcentajes = plt.pie(
    df["Dinero recibido"],
    labels=df["Club"],
    colors=colores,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor':'black'}
)

# Configurar color de cada porcentaje
porcentajes[0].set_color('black')   # Colo Colo
porcentajes[1].set_color('white')   # U
porcentajes[2].set_color('white')   # Católica
porcentajes[3].set_color('black')   # Otros

# Hacer porcentajes más visibles
for p in porcentajes:
    p.set_fontsize(12)
    p.set_weight('bold')

plt.title(
    "Distribución de ingresos por derechos de transmisión",
    fontsize=14,
    fontweight='bold'
)

plt.axis('equal')
plt.show()
```
