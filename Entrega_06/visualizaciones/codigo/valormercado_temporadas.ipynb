```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from ipywidgets import interact

# 1. Cargar y preparar los datos
df = pd.read_excel('/content/Base de datos Patricio Gómez (Limpia).xlsx')

# FILTRO CRUCIAL: Limitar los datos estrictamente hasta la temporada 2025
df = df[df['Temporada'] <= 2025]

df_grouped = df.groupby(['Temporada', 'Equipo'])['Valor de mercado'].sum().reset_index()

# Función para asignar las categorías de colores institucionales
def asignar_categoria(nombre_equipo):
    nombre_lower = str(nombre_equipo).lower()
    if 'colo' in nombre_lower: return 'Colo Colo'
    elif 'chile' in nombre_lower: return 'U. de Chile'
    elif 'católica' in nombre_lower or 'catolica' in nombre_lower: return 'U. Católica'
    else: return 'Otros'

df_grouped['Equipo_Categoria'] = df_grouped['Equipo'].apply(asignar_categoria)

# Definir la paleta de colores fija
paleta_colores = {
    'Colo Colo': 'black',
    'U. de Chile': '#0047AB',  # Azul eléctrico
    'U. Católica': '#87CEEB', # Celeste
    'Otros': '#D3D3D3'         # Gris claro
}

# Obtener la lista de temporadas disponibles (acotadas por el filtro anterior)
temporadas = sorted(df_grouped['Temporada'].unique())

# 2. Crear la función que dibuja el gráfico para una temporada específica
def graficar_temporada(temporada):
    # Filtrar datos por el año seleccionado
    df_año = df_grouped[df_grouped['Temporada'] == temporada].copy()
    # Ordenar de mayor a menor según el valor de mercado
    df_año = df_año.sort_values(by='Valor de mercado', ascending=False)

    # Configurar el tamaño y estilo del gráfico
    plt.figure(figsize=(10, 6))
    sns.set_style("whitegrid")

    # Crear el gráfico de barras horizontales
    ax = sns.barplot(
        data=df_año,
        x='Valor de mercado',
        y='Equipo',
        hue='Equipo_Categoria',
        palette=paleta_colores,
        dodge=False,
        legend=True
    )

    # Títulos y etiquetas
    plt.title(f'Ranking de Valor de Mercado - Temporada {temporada}', fontsize=14, pad=15, weight='bold')
    plt.xlabel('Valor de Mercado Total', fontsize=11)
    plt.ylabel('Equipo', fontsize=11)

    # Ajustar límites del eje X para que no se mueva bruscamente al cambiar de año
    plt.xlim(0, df_grouped['Valor de mercado'].max() * 1.1)

    # Mover la leyenda fuera del gráfico para que no estorbe
    sns.move_legend(ax, "upper right", bbox_to_anchor=(1.25, 1), title='Categorías')

    plt.tight_layout()
    plt.show()

# 3. Crear el menú interactivo desplegable de Google Colab
interact(graficar_temporada, temporada=widgets.Dropdown(
    options=temporadas,
    value=temporadas[0],
    description='Temporada:',
    disabled=False,
))
```
