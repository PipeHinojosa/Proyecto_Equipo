```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from ipywidgets import interact

# 1. Leer el archivo CSV especificando la codificación y el detector automático de separador
df = pd.read_csv('dt_cambios_temporadas_2018_2025.csv', sep=None, engine='python', encoding='latin1')

# FILTRO DE TEMPORADAS: Restringir el DataFrame únicamente entre 2019 y 2025
df = df[(df['Temporada'] >= 2019) & (df['Temporada'] <= 2025)].copy()

# Obtener la lista de todas las temporadas disponibles después del filtro
temporadas_disponibles = sorted(df['Temporada'].unique())

# DETERMINACIÓN DINÁMICA: Tomamos el último año disponible en la lista de forma automática
ultima_temporada = temporadas_disponibles[-1]

# 2. Definir la función que dibuja el gráfico para la temporada elegida
def graficar_temporada(temporada_seleccionada):
    # Filtrar por el año elegido y ordenar por posición de mejor a peor (ascendente)
    df_filtrado = df[df['Temporada'] == temporada_seleccionada].copy()
    df_filtrado = df_filtrado.sort_values(by='Posición final torneo', ascending=True)

    # Configurar fondo BLANCO y estilo limpio
    plt.figure(figsize=(10, 8))
    sns.set_style("whitegrid")
    plt.rcParams['figure.facecolor'] = '#ffffff'
    plt.rcParams['axes.facecolor'] = '#ffffff'

    # Crear el gráfico de barras horizontales usando EQUIPO en el Eje Y
    ax = sns.barplot(
        data=df_filtrado,
        x='Cantidad de técnicos',
        y='Equipo',
        orient='h',
        color='#2ca02c',  # Verde estándar
        alpha=0.85,
        order=df_filtrado['Equipo'] # Se ordenan en base a cómo quedaron en la tabla
    )

    # Colocar la posición final del torneo al costado derecho de cada barra
    for i, (index, fila) in enumerate(df_filtrado.iterrows()):
        pos_x = fila['Cantidad de técnicos'] + 0.05
        texto_posicion = f"({int(fila['Posición final torneo'])}°)"
        ax.text(
            pos_x, i, texto_posicion,
            va='center', ha='left',
            color='black', fontsize=11, fontname='DejaVu Sans', weight='bold'
        )

    # Formatear títulos y etiquetas en color NEGRO
    plt.title(f'Temporada {temporada_seleccionada}: Cantidad de técnicos por Equipo',
              fontsize=14, color='black', pad=20, weight='bold')
    plt.xlabel('Cantidad de técnicos durante la temporada', fontsize=11, color='black', labelpad=10)
    plt.ylabel('Equipos (Ordenados por Posición Final: Mejor a Peor)', fontsize=11, color='black', labelpad=10)

    # FORZAR NÚMEROS ENTEROS EN EL EJE X
    max_tecnicos = int(df['Cantidad de técnicos'].max())
    plt.xticks(range(0, max_tecnicos + 2), color='black') # Rango estricto: 0, 1, 2, 3...

    plt.yticks(color='black')
    plt.xlim(0, max_tecnicos + 1.2)  # Darle margen a la etiqueta del texto para que no se corte

    # Ajustar las rejillas para mayor limpieza visual
    ax.grid(True, axis='x', linestyle='--', alpha=0.5, color='#ccc')
    ax.grid(False, axis='y')  # Apagar rejilla horizontal

    # Limpiar bordes innecesarios
    sns.despine(left=True, bottom=True)
    plt.tight_layout()
    plt.show()

# 3. Desplegar el control interactivo en pantalla
interact(graficar_temporada, temporada_seleccionada=widgets.Dropdown(
    options=temporadas_disponibles,
    value=ultima_temporada, # <--- AHORA ES DINÁMICO: Carga automáticamente el año más reciente de la lista
    description='Temporada:',
    disabled=False,
))
```
