```python
import pandas as pd
import altair as alt

# 1. Leer el archivo CSV especificando la codificación y el separador correcto
df = pd.read_csv('dt_cambios_temporadas_2018_2025.csv', sep=None, engine='python', encoding='latin1')

# 2. Filtrar los datos únicamente para la temporada 2025
df_2025 = df[df['Temporada'] == 2025].copy()

# TRUCO VISUAL: Si es 0, le damos un ancho mínimo de 0.08 para que dibuje una barra pequeña
df_2025['Ancho_Barra'] = df_2025['Cambios de técnico'].apply(lambda x: 0.08 if x == 0 else x)

# 3. Crear una columna de texto personalizada para mostrar al costado de las barras
df_2025['Etiqueta'] = df_2025['Equipo'] + ' (' + df_2025['Posición final torneo'].astype(str) + '°)'

# 4. Definir el gráfico utilizando 'Ancho_Barra' para el eje X
base = alt.Chart(df_2025).encode(
    x=alt.X('Ancho_Barra:Q',
            title='Cantidad de cambios de director técnico',
            axis=alt.Axis(format='d', tickMinStep=1, values=[0, 1, 2, 3])), # Forzamos los números exactos en el eje
    y=alt.Y('Posición final torneo:O',
            title='Posición Final en el Torneo (Mejor a Peor)',
            sort='ascending'),
    # El tooltip sigue mostrando el valor real original de la base de datos
    tooltip=['Equipo', alt.Tooltip('Cambios de técnico:Q', title='Cambios de técnico'), 'Posición final torneo']
).properties(
    width=600,
    height=500,
    title='Temporada 2025: Relación entre cambios de DT y posición final'
)

# 5. Combinar las barras con las etiquetas de texto
grafico_final = base.mark_bar(color='#39ff14', opacity=0.8) + base.mark_text(
    align='left',
    dx=6,
    fontSize=11,
    font='Segoe UI'
).encode(
    text='Etiqueta:N'
)

# 6. Desplegar el gráfico en la pantalla de Colab
grafico_final
```
